# Del 2: `for`-løkker og funksjoner 🔂

En **liste** er flere verdier inne i firkantklammer: `[2, 5, 7]`.
Det eneste vi trenger å vite om lister, er at vi kan **gå gjennom** dem med en `for`-løkke:

```python
for x in [2, 5, 7]:
    dette gjentas én gang for hver verdi
```

`x` får verdiene **én etter én**: først `2`, så `5` og til slutt `7`.
Alt som har innrykk, kjøres én gang for hver verdi.

## 2.1 Gå gjennom en liste

````python
for x in [2, 5, 7]:
    print("x er", x)

print("Ferdig!")
````

🤔 Hvor mange ganger skrives `Ferdig!` ut? Hvorfor?

````python
for x in [1, 2, 3, 4, 5]:
    print(x, "i andre er", x**2)
````

````python
# Verditabell for f(x) = x² - 2x
for x in [-2, -1, 0, 1, 2, 3, 4]:
    y = x**2 - 2*x
    print(x, y)
````

💡 Linjen `y = x**2 - 2*x` ser ut som likningen til grafen, men den er en **utregning** som skjer én gang per runde.
Resultatet er en tabell med **punkter** på grafen, ikke selve grafen.

Lista kan også lagres i en variabel. Navnet på løkkevariabelen kan vi velge selv:

````python
temperaturer = [-10, 0, 20, 37, 100]

for c in temperaturer:
    F = c * 9/5 + 32
    print(c, "grader Celsius er", F, "grader Fahrenheit")
````

## 2.2 Funksjoner: `def f(x)` 📈

I matematikk skriver vi $f(x) = x^2 - 2x$. I Python skriver vi:

````python
def f(x):
    return x**2 - 2*x

print(f(3))
print(f(0), f(10))
````

* `def` betyr at vi **definerer** en funksjon. Den heter `f` og har variabelen `x`
* `return` gir svaret **tilbake**
* Når vi skriver `f(3)`, regner Python ut funksjonsuttrykket med `x = 3`

Funksjonen kan hete hva som helst, for eksempel `A(x)` for areal eller `O(x)` for overskudd.

````python
# Verditabell med en funksjon
def f(x):
    return x**2 - 2*x

for x in [-2, -1, 0, 1, 2, 3, 4]:
    print(x, f(x))
````

### ✨ Penere utskrift med f-streng

Setter vi `f` foran anførselstegnet, blir alt inni `{ }` regnet ut og satt inn i teksten:

````python
def f(x):
    return x**2 - 2*x

for x in [-2, -1, 0, 1, 2, 3, 4]:
    print(f"f({x}) = {f(x)}")
````

## 2.4 Summere med en løkke ⭐

Dette er **det viktigste mønsteret** i kurset, og det dukker opp på eksamen hele tiden:

1. Start med `S = 0`
2. Legg til noe i hver runde: `S = S + ...`
3. Skriv ut `S` **etter** løkka

````python
# 1 + 2 + 3 + 4 + 5
S = 0
for i in range(1, 6):
    S = S + i
print(S)
````

Slik endrer variablene seg (en **sporingstabell**):

| Runde | `i` | `S` før | `S` etter |
|---|---|---|---|
| 1 | 1 | 0 | 1 |
| 2 | 2 | 1 | 3 |
| 3 | 3 | 3 | 6 |
| 4 | 4 | 6 | 10 |
| 5 | 5 | 10 | 15 |

Legg merke til at `i` og `S` aldri har flere verdier *samtidig*. De får nye verdier **etter hverandre**, én runde om gangen.

````python
# 🤔 Hva er forskjellen? print har fått innrykk!
S = 0
for i in range(1, 6):
    S = S + i
    print(S)
````

````python
# 1 + 2 + ... + 100
S = 0
for i in range(1, 101):
    S = S + i
print(S)
````

````python
# Summen av kvadrattall: 1² + 2² + ... + 10²
S = 0
for i in range(1, 11):
    S = S + i**2
print(S)
````

Vi kan også skrive `S += i` i stedet for `S = S + i`. Det betyr det samme.

````python
# Aritmetisk rekke: 3 + 7 + 11 + 15 + ...  (a1 = 3, d = 4)
a = 3
d = 4
N = 10
S = 0

for i in range(N):
    S = S + a
    a = a + d

print(S)
````

Dette er en gammel eksamensoppgave! 🤔 Hva blir svaret hvis `N = 100`? Sjekk med formelen $S_n = \frac{n(a_1 + a_n)}{2}$.

````python
# Produkt: 5! = 1 * 2 * 3 * 4 * 5
P = 1
for i in range(1, 6):
    P = P * i
print(P)
````

🤔 Hvorfor starter vi med `P = 1` og ikke `P = 0`?

````python
# Renter: 10 000 kr med 5 % rente i 10 år
belop = 10000
for aar in range(1, 11):
    belop = belop * 1.05
    print("Etter", aar, "år:", round(belop, 2), "kr")
````

## 2.5 `for` og `if` sammen

Vi kan ha en `if` inne i en løkke. Da sjekkes betingelsen én gang for hver verdi.

````python
# For hvilke heltall x mellom -10 og 10 er f(x) = x² - 3x - 4 negativt?
def f(x):
    return x**2 - 3*x - 4

antall = 0
for x in range(-10, 11):
    if f(x) < 0:
        print(x, "gir negativ verdi")
        antall = antall + 1
print("Antall:", antall)
````

## 2.6 Desimaltall og steglengde ⭐

`range` gir bare heltall. Trenger vi steg på for eksempel 0.1, bruker vi dette trikset (som også står i mange eksamensoppgaver):

```python
x = start + i*h
```

````python
# Verditabell for f(x) = x² fra 0 til 1 med steg 0.1
def f(x):
    return x**2

start = 0
h = 0.1

for i in range(11):
    x = start + i*h
    print(round(x, 1), round(f(x), 3))
````

## 2.7 Areal under en graf (rektangelsum) ⭐

Vi deler intervallet $[a, b]$ i `n` like rektangler med bredde $h = \frac{b-a}{n}$
og legger sammen arealene: høyde × bredde = $f(x) \cdot h$.

````python
# Areal under f(x) = x² fra 0 til 1
def f(x):
    return x**2

a = 0
b = 1
n = 10

h = (b - a) / n
S = 0
for i in range(n):
    x = a + i*h
    S = S + f(x) * h

print(S)
````

# Del 3: `while`, gjenta så lenge noe er sant 🔁

* `for`: vi vet **hvor mange ganger** noe skal gjentas
* `while`: vi gjentar **så lenge** betingelsen er sann, og vet ofte *ikke* på forhånd hvor mange ganger det blir

```python
while betingelse:
    gjenta dette
```

⚠️ Noe inne i løkka **må endre seg** slik at betingelsen til slutt blir usann.
Ellers får vi en **evig løkke**. Den stopper du med ■-knappen ved cellen.

## 3.1 Første eksempler

````python
n = 5
while n > 0:
    print(n)
    n = n - 1
print("Takeoff! 🚀")
````

Det samme kan gjøres med både `for` og `while`:

````python
for i in range(1, 6):
    print(i)
````

````python
i = 1              # 1. startverdi
while i <= 5:      # 2. betingelse
    print(i)
    i = i + 1      # 3. oppdatering
````

## 3.2 Når vet vi ikke antall ganger?

Her er `while` den riktige løkka.

````python
# Hvor mange år tar det før 10 000 kr har doblet seg med 5 % rente?
belop = 10000
aar = 0

while belop < 20000:
    belop = belop * 1.05
    aar = aar + 1

print("Det tar", aar, "år")
print("Beløpet er da", round(belop, 2), "kr")
````

🤔 Endre renten til 10 %. Hvor lang tid tar det nå?

````python
# Et legemiddel: 100 mg i blodet, og 20 % forsvinner hver time.
# Hvor lang tid tar det før det er under 10 mg igjen?
mengde = 100
timer = 0

while mengde >= 10:
    mengde = mengde * 0.8
    timer = timer + 1

print("Etter", timer, "timer er det", round(mengde, 2), "mg igjen")
````

````python
# Hvor mange ledd i 1 + 2 + 3 + ... trenger vi for at summen skal bli over 1000?
n = 0
S = 0

while S <= 1000:
    n = n + 1
    S = S + n

print("Vi trenger", n, "ledd. Summen er da", S)
````

````python
# Brukeren må skrive et positivt tall
tall = float(input("Skriv et positivt tall: "))

while tall <= 0:
    tall = float(input("Det var ikke positivt. Prøv igjen: "))

print("Kvadratroten er", tall ** 0.5)
````

## 4.1 Plassholder-x

````python
# 🤔 Hva skrives ut?
def f(x):
    return x**2

x = 3
print(f(5))
print(f(x))
````

## 4.2 Gjennomsnittlig vekstfart

Gjennomsnittlig vekstfart fra $x = a$ til $x = b$ er stigningstallet til linja gjennom $(a, f(a))$ og $(b, f(b))$:

$$\frac{f(b) - f(a)}{b - a}$$

````python
# Gjennomsnittlig vekstfart fra x = 0 til x = 5 (eksamensoppgave: Ada)
def f(x):
    return x**2 - 3*x + 7

a = 0
b = 5
v = (f(b) - f(a)) / (b - a)
print(v)
````

## 4.3 Momentan vekstfart med liten h

Lar vi intervallet bli veldig lite, $b = a + h$ med en liten $h$, blir gjennomsnittlig vekstfart nesten lik den **momentane** vekstfarten $f'(a)$:

$$f'(a) \approx \frac{f(a + h) - f(a)}{h}$$

````python
# Momentan vekstfart i x = 2 (tilnærmet f'(2))
def f(x):
    return x**2 - 3*x + 7

a = 2
h = 0.0001
v = (f(a + h) - f(a)) / h
print(v)
````

🤔 Prøv med `h = 1`, `h = 0.1` og `h = 0.01`. Hvilket tall nærmer svaret seg? Sjekk med derivasjon.

````python
# En funksjon kan bruke en annen funksjon
def f(x):
    return x**2 - 3*x + 7

h = 0.0001

def df(x):
    return (f(x + h) - f(x)) / h

print(df(2))
print(df(1.5))
````

## 4.4 Finne toppunkt med `while` ⭐⭐

**Strategien:** start til venstre og gå små steg mot høyre **så lenge grafen stiger**,
altså så lenge $f(x+h) > f(x)$. Når grafen slutter å stige, er vi på toppen.

Merk:
* `x` i `def f(x)` er plassholderen, mens `x = 0` er en boks som flytter seg mot høyre, ett steg om gangen
* Svaret blir **omtrent** riktig. Hvor nøyaktig det blir, avhenger av steglengden `h`

````python
def f(x):
    return -x**2 + 4*x + 1

x = 0
h = 0.01

while f(x + h) > f(x):
    x = x + h

print("Toppunkt omtrent i x =", round(x, 2), "med f(x) =", round(f(x), 2))
````

### Toppunkt med den deriverte

På eksamen ser toppunktprogrammet ofte slik ut. Strategien er den samme:
vi går mot høyre **så lenge den deriverte er positiv**, altså så lenge grafen stiger.

````python
def f(x):
    return -x**2 + 4*x + 1

h = 0.0001

def df(x):
    return (f(x + h) - f(x)) / h

x = 0
while df(x) > 0:
    x = x + 0.01

print("Toppunkt omtrent i x =", round(x, 2))
````

# Oppsummering 📝

| Verktøy | Brukes når | Eksempel |
|---|---|---|
| `if` / `elif` / `else` | vi skal **velge** hva som skjer | fortegn, diskriminant |
| `for x in [...]` | vi skal gå gjennom en **liste** | verditabell |
| `for i in range(...)` | vi vet **hvor mange ganger** | sum, rekke, rektangelsum |
| `while betingelse:` | vi gjentar **så lenge** noe er sant | doblingstid, toppunkt |
| `S = S + ...` | vi skal **summere** | $1 + 2 + \dots + n$ |
| `def f(x): return ...` | vi trenger en **matematisk funksjon** | $f(x) = x^2 - 3x$ |

**Husk:** kolon `:` på slutten av linjen og **innrykk** på det som hører til!
