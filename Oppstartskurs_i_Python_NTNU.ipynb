````python
# if/elif/else

dog_says = input("Dog says: ")

if dog_says == "awo!":
  print("There there, it's ok..")
elif dog_says == "bark!":
  print("how rude!")
else:
  print("Sorry, I don't speak dog.")
````

````python
# for løkke

dog_breeds = ["Samoyed", "Border Collie", "Husky"]

for breed in dog_breeds:
  print(breed, "is the best!")
````

## 🔁 3. Funksjoner som kaller andre funksjoner


````python
# Eksempel: enkel kalkulator som bruker funksjoner

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def calculator(a, b, operation):
    if operation == "add":
        return add(a, b)
    elif operation == "subtract":
        return subtract(a, b)
    else:
        return "Ukjent operasjon"

print(calculator(10, 5, "add"))
print(calculator(10, 5, "subtract"))
````
